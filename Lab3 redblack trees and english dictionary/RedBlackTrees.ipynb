```python
class Node:
    # Node class constructor 
    def __init__(self, data):
        self.data = data              # Value stored in node
        self.color = "RED"            # New nodes are inserted as RED
        self.left = None              # Left child
        self.right = None             # Right child
        self.parent = None            # Parent pointer


class RedBlackTree:
    # class constructor for Red-Black Tree
    def __init__(self):
        self.NIL = Node(None)        # Create a special NIL node instead of using None
        self.NIL.color = "BLACK"     # NIL nodes are always BLACK 
        self.NIL.left = self.NIL     # NIL points to itself
        self.NIL.right = self.NIL    # NIL points to itself
        self.root = self.NIL         # Initially, tree is empty
        self.size = 0                # Number of nodes in tree
    
    #---------------------------------------------------------------------------
    # wrapper method to search for a key in the tree
    def search(self, key):
        return self._search(self.root, key)
    # helper method for search 
    def _search(self, node, key):
        # Base case node not found or key matches
        if node is self.NIL or key == node.data: 
            return node
        # Traverse left or right based on comparison
        if key < node.data:
            return self._search(node.left, key)
        return self._search(node.right, key)

    #---------------------------------------------------------------------------
    # left-rotation around node x
    def left_rotate(self, x):
        """
        Perform left rotation around node x

            x                   y
             \                 / \
              y      ==>      x   T3
             / \               \
           T2  T3              T2
        """
        y = x.right                 # y becomes new root of this subtree
        x.right = y.left            # Turn y's left subtree into x's right subtree
        # Update parent pointer of T2 if it exists
        if y.left is not self.NIL:  
            y.left.parent = x
        y.parent = x.parent        
        # If x was root, update root
        if x.parent is None:
            self.root = y
        elif x == x.parent.left:
            x.parent.left = y
        else:
            x.parent.right = y
        y.left = x                  # Put x as left child of y
        x.parent = y                # Update parent pointer

    #---------------------------------------------------------------------------
    # right-rotation around node y
    def right_rotate(self, y):
        """
        Perform right rotation around node y.

              y                 x
             /                 / \
            x       ==>       T1  y
           / \                   /
         T1  T2                T2
        """
        x = y.left
        y.left = x.right
        # Update parent pointer of T2
        if x.right is not self.NIL: 
            x.right.parent = y
        x.parent = y.parent
        # Update root if y was root
        if y.parent is None:
            self.root = x
        elif y == y.parent.right:
            y.parent.right = x
        else:
            y.parent.left = x
        x.right = y
        y.parent = x

    #---------------------------------------------------------------------------
    # insert a new key 
    def insert(self, key):
        """
        Insert a new node into the Red-Black Tree.
        """
        node = Node(key)
        node.left = self.NIL       
        node.right = self.NIL
        parent = None
        current = self.root

        
        while current is not self.NIL: 
            parent = current
            if node.data < current.data:
                current = current.left
            elif node.data > current.data:
                current = current.right
            else:
                return False # No duplicates allowed
        node.parent = parent

        # If tree is empty → new node is root
        if parent is None:
            self.root = node
        elif node.data < parent.data:
            parent.left = node
        else:
            parent.right = node
        # increment the size variable
        self.size += 1
        # Fix Red-Black Tree violations
        self.fix_insert(node)
        return True

    #---------------------------------------------------------------------------
    # fix the tree after insertion to maintain Red-Black properties
    def fix_insert(self, z):
        """
        Restore Red-Black properties after insertion.
        Handles 3 main cases:
        1. Recoloring
        2. Rotation (Left)
        3. Rotation (Right)
        """
        while z.parent and z.parent.color == "RED":
            # Case: Parent is left child of grandparent
            if z.parent == z.parent.parent.left:
                uncle = z.parent.parent.right

                # Case when Uncle is RED → recolor
                if uncle.color == "RED":
                    z.parent.color = "BLACK"
                    uncle.color = "BLACK"
                    z.parent.parent.color = "RED"
                    z = z.parent.parent   # to check for violations up the tree

                # Case when Uncle is BLACK → rotations needed
                else:
                    if z == z.parent.right:
                        z = z.parent
                        self.left_rotate(z)
                    z.parent.color = "BLACK"
                    z.parent.parent.color = "RED"
                    self.right_rotate(z.parent.parent)

            # case: parent is right child of grandparent
            else:
                uncle = z.parent.parent.left

                if uncle.color == "RED":
                    z.parent.color = "BLACK"
                    uncle.color = "BLACK"
                    z.parent.parent.color = "RED"
                    z = z.parent.parent

                else:
                    if z == z.parent.left:
                        z = z.parent
                        self.right_rotate(z)

                    z.parent.color = "BLACK"
                    z.parent.parent.color = "RED"
                    self.left_rotate(z.parent.parent)
        # Ensure root is always BLACK even afer fixes in case it becomes RED
        self.root.color = "BLACK"

    # ------------------------------------------------------------------------------------
    # wrapper method to calculate the height of the tree
    def height(self):
        """Returns height = longest path from root to leaf."""
        return self._height(self.root)
    # helper method to calculate the height of the tree
    def _height(self, node):
        if node is self.NIL: 
            return 0
        return 1 + max(self._height(node.left), self._height(node.right))
    
    #-----------------------------------------------------------------------------------
    # claculate the black height of the tree
    def black_height(self):
        """Count number of BLACK nodes from root to any leaf."""
        node = self.root
        height = 0
        # follow any path and it will count the same number of BLACK nodes
        while node is not self.NIL:  
            if node.color == "BLACK":
                height += 1
            node = node.left
        return height
    #----------------------------------------------------------------------------------
    # return the number of nodes in the tree
    def get_size(self):
        """return the number of nodes in the tree"""
        return self.size
```

```python
# small test to check the implementation
rbt = RedBlackTree()
values = [10, 20, 30, 15, 25, 5]
for v in values:
    rbt.insert(v)
# checking for existing and non-existing keys
print("Search 15:", rbt.search(15).data if rbt.search(15) != rbt.NIL else "Not Found")
print("Search 100:", rbt.search(100).data if rbt.search(100) != rbt.NIL else "Not Found")
print("Tree Height:", rbt.height())
print("Black Height:", rbt.black_height())
print("Tree Size:", rbt.get_size())
```

```python
# dictionary interface using the Red-Black Tree for storage
class EnglishDictionary:
    # class constructor 
    def __init__(self, filename):
        self.filename = filename
        self.tree = RedBlackTree()
        self.load_dictionary()

    # load the dictionary from the file into the Red-Black Tree
    def load_dictionary(self):
        try:
            with open(self.filename, "r") as f:
                for line in f:
                    word = line.strip().lower()
                    if word:
                        self.tree.insert(word)
        except FileNotFoundError:
            # Create the file if it doesn't exist
            open(self.filename, "w").close()

    # insert a new word into the dictionary and update the file
    def insert_word(self, word):
        word = word.strip().lower()
        if self.tree.insert(word):
            with open(self.filename, "a") as f:
                f.write(word + "\n")
            print(f"Word '{word}' added to dictionary.")
        else:
            print("ERROR: Word already in the dictionary!")

    # look-up a word in the dictionary and print YES or NO
    def lookup_word(self, word):
        word = word.strip().lower()
        if self.tree.search(word) is not self.tree.NIL:  
            print("YES")
        else:
            print("NO")
```

```python
dictionary = EnglishDictionary("dictionary.txt")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
def visualize_tree(tree):
    fig, ax = plt.subplots(figsize=(12, 8))
    ax.set_aspect('equal')
    ax.axis('off')

    def draw_node(node, x, y, dx):
        if node is tree.NIL:
            return
        # Draw edges to children first
        if node.left is not tree.NIL:
            ax.plot([x, x - dx], [y, y - 1.5], 'k-', linewidth=1)
            draw_node(node.left, x - dx, y - 1.5, dx / 2)
        if node.right is not tree.NIL:
            ax.plot([x, x + dx], [y, y - 1.5], 'k-', linewidth=1)
            draw_node(node.right, x + dx, y - 1.5, dx / 2)

        # Draw the node circle with its color
        color = "#c0392b" if node.color == "RED" else "#2c2c2c"
        circle = mpatches.Circle((x, y), 0.4, color=color, zorder=3)
        ax.add_patch(circle)
        ax.text(x, y, str(node.data), ha='center', va='center',
                color='white', fontsize=8, fontweight='bold', zorder=4)
    draw_node(tree.root, x=0, y=0, dx=3)
    plt.tight_layout()
    plt.show()

visualize_tree(dictionary.tree)
```
